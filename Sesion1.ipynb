```python
a = 12.5
b = 14.2
c = a + b
print(c)

# Operadores con número
a = 12.5
b = 14.2
print("Suma:", a+b)
print("Resta:", b-a)

# Promedio
print("Promedio de dos lecturas", (a+b)/2)

# Comparaciones
# > mayor que
# < menor que
#  == igual que
print("¿a es mayor que b?", a > b)
print("¿a es menor que b?", a < b)
print("¿a es igual que b?", a == b)

```
